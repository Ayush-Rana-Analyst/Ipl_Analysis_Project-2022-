```python
5.Key Player Performance
 1.Player of the Match
```

```python
count=df['player_of_the_match'].value_counts().head(10)
sns.barplot(x=count.values,y=count.index,palette='mako')
plt.title('Top 10 players of Man of the Match')
```

```python
2. Best Bowling Figure 
```

```python
# df['best_bowling_figure'].
df['highest_wicket']=df['best_bowling_figure'].apply(lambda x:x.split('--')[0])
df['highest_wicket']=df['highest_wicket'].astype(int)
df['highest_wicket']

best_bowlers=df.groupby('best_bowling')['highest_wicket'].sum().sort_values(ascending=False).head(10)
figure=plt.figure(figsize=(9,5))
sns.barplot(x=best_bowlers.values,y=best_bowlers.index)
plt.title('Best bowelers in IPL')
    
```
